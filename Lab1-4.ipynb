### 2. Pandas

Найдите ответы на вопросы или выполните действия с предложенным датасетом 

Файл: `russian-housing-market.csv`

```
price_doc - Цена продажи
id - Id транзакции
timestamp - Дата транзакции
full_sq - суммарная площадь в квадратных метрах, включая лоджию, валконы и отдельные нежилые помещения
life_sq - жилая площаль в квадратных метрах, исключая лоджии, балконы и прочие нежилые помещения
floor - для апартаментов, этаж помещения
max_floor - количество этажей в здании

product_type - покупка или инвестиция собственника-арендатора
sub_area - название района
```

````python
import pandas as pd
````

````python
# Load dataset here
df = pd.read_csv('russian-housing-market.csv')
````

1. Посчитайте количество записей в датасете с отсутствующей информацией о жилой площади жилого помещения

````python
ans = df['life_sq'].isna().sum()
print(ans)
````

2. Предложите способ заменить отсутствующие значения из пункта 1 и реализуйте его. Способ нужно аргументировать(сказать почему это достаточно хорошее приближение)

````python
medi = (df['life_sq'] / df['full_sq']).median()
df['life_sq'] = df['life_sq'].fillna(df['full_sq'] * medi)

print( df['life_sq'].isna().sum())
# Данный были заполнены исходя из медианного соотношения жилой площади к общей площади. 
# Так мы учитываем размер квартиры, не занижаем значения жилой площади для больших квартир и не завышаем для маленьких.
````

3. Какие значения имеет 25, 50 и 75 перцентили полной площади помещения

````python
ans = df['full_sq'].quantile([0.25, 0.50, 0.75])
print(ans)
````

4. Посчитайте среднюю стоимость продажи помещений с общей площадью от 50 до 70 квадратных метров

````python
cut = df[(df['full_sq'] <= 70) & (df['full_sq'] >= 50)]
ans = cut['price_doc'].mean()
print(ans)
````

5. Какая максимальная этажность здания представлена в датасете

````python
ans = df['max_floor'].max()
print(ans)
````

6. Выявите район с самым дорогим жильем

````python
ans = df.groupby('sub_area')['price_doc'].mean().idxmax()
print(ans)
````

7. Посчитайте среднюю стоимость за квадратный метр жилья в Митино(Mitino)

````python
mi = df[df['sub_area'] == 'Mitino'].copy()
mi['price_per_sq'] = mi['price_doc'] / mi['full_sq']
ans = mi['price_per_sq'].mean()
print(ans)
````

8. Посчитайте среднюю площадь комнаты в Бибирево

````python
bi = df[df['sub_area'] == 'Bibirevo'].copy()
bi['room_area'] = bi['life_sq'] / bi['num_room']
ans = bi['room_area'].mean()
print(ans)
````

9. Проанализируйте есть ли разница в цене между квартирами на 1 и 9 этажах

````python
floor1 = df[df['floor'] == 1]['price_doc'].mean()
floor9 = df[df['floor'] == 9]['price_doc'].mean()
print("floor1:", floor1, "floor9:", floor9, "difference:", np.abs(floor1 - floor9))
````

10. Найдите минимальную стоимость квартиры с жилой площадью не менее 60 квадратных метров, не менее чем 3мя комнатами и не ниже чем на 5 этаже

````python
cut = df[(df['life_sq'] >= 60) & (df['num_room'] >= 3) & (df['floor'] >= 5)]
print(cut['price_doc'].min())
````

### 3. Визуализация

Используя библиотеки matplotlib/sns/plotly/pandas и набор данных представленный ранее визуализируйте

````python
import matplotlib.pyplot as plt
````

1. Как менялась средняя цена квартиры в Нагатинском Затоне(Nagatinski Zaton) с течением времени

````python
area = df[df['sub_area'] == 'Nagatinskij Zaton']
area['timestamp'] = pd.to_datetime(area['timestamp'])
monthly_price = area.groupby('timestamp')['price_doc'].mean()
plt.figure(figsize=(12, 6))
plt.plot(monthly_price.index, monthly_price.values)

plt.title('Средняя цена квартир по месяцам')
plt.xlabel('Месяц')
plt.ylabel('Средняя цена')
plt.xticks(rotation=45)
plt.grid()
plt.tight_layout()
plt.show()
````

2. Постройте гистограмму распределения цен на квартиры с общей площадью более 50 кв.метров

````python
prices = df[df['full_sq'] > 50]['price_doc'].dropna()

plt.hist(prices, bins=50)

plt.title('Распределение цен квартир площадью больше 50 м²')
plt.xlabel('Цена квартиры')
plt.ylabel('Количество квартир')
plt.grid(axis='y')
plt.tight_layout()
plt.show()
````

3. Визуализируйте количество проданных квартир в разных районах

````python
districts = df['sub_area'].value_counts().sort_values()

plt.figure(figsize=(14, 25))
plt.barh(districts.index, districts.values)

plt.title('Количество квартир по районам')
plt.xlabel('Количество квартир')
plt.ylabel('Район')
plt.tight_layout()
plt.show()
````

4. Постройте box plot отображающий распределение полных площадей квартир в Басманном районе

````python
area = df[df['sub_area'] == 'Basmannoe']['full_sq'].dropna()

plt.boxplot(area)

plt.title('Распределение площадей квартир в Басманном районе')
plt.ylabel('Площадь квартиры (м²)')
plt.grid(axis='y')
plt.tight_layout()
plt.show()
````

5. Визуализируйте зависимость средней стоимости квартир в зависимости от этажа

````python
floor_price = df.groupby('floor')['price_doc'].mean().dropna()

plt.plot(floor_price.index, floor_price.values)

plt.title('Средняя стоимость квартир в зависимости от этажа')
plt.xlabel('Этаж')
plt.ylabel('Средняя цена')
plt.grid()
plt.tight_layout()
plt.show()
````

6. Покажите как меняется средняя площадь квартир в зависимости от района

````python
district_area = df.groupby('sub_area')['full_sq'].mean().sort_values()

plt.figure(figsize=(14, 25))
plt.barh(district_area.index, district_area.values)

plt.title('Средняя площадь квартир по районам')
plt.xlabel('Средняя площадь (м²)')
plt.ylabel('Район')
plt.tight_layout()
plt.show()
````

7. Постройте график изменения медианной цены на квартиру менее 50 кв. метров за период с 2011 до 2014 года*

\* Задание на доп. балл

````python
sflats= df[(df['full_sq'] < 50) & (df['timestamp'] >= '2011-01-01') & (df['timestamp'] < '2015-01-01')]

monthly_area = sflats.set_index('timestamp')['full_sq'].resample('ME').median()

plt.plot(monthly_area.index, monthly_area.values, marker='o')

plt.title('Медианная площадь небольших квартир по месяцам (2011–2014)')
plt.xlabel('Месяц')
plt.ylabel('Медианная площадь (м²)')
plt.xticks(rotation=45)
plt.grid()
plt.tight_layout()
plt.show()
````
